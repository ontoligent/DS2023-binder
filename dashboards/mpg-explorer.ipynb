# Fuel Economy, 1970–1982

These data describe 392 cars sold in the United States between 1970 and 1982, built in the USA, Europe, and Japan.

Use the controls under each plot to compare cars from the three regions.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ipywidgets import interact

MPG = sns.load_dataset('mpg').dropna()
measure_cols = ['mpg', 'horsepower', 'weight', 'acceleration', 'displacement']
```

## How do the regions differ?

Pick a measure and a number of bins.

Each region gets its own histogram.

```python
@interact(measure=measure_cols, bins=(5, 50, 5))
def plot_hist(measure='mpg', bins=20):
    (MPG
        .pivot(columns='origin', values=measure)
        .plot.hist(bins=bins, alpha=0.5, figsize=(6, 4), title=measure))
    plt.show()
```

American cars are heavier and more powerful, and they get fewer miles per gallon.

Japanese and European cars are much closer to each other than either is to American cars.

Which measure separates the regions best?

## How did cars change over time?

Pick a measure to see its average for each model year.

```python
@interact(measure=measure_cols)
def plot_trend(measure='mpg'):
    (MPG
        .groupby(['model_year', 'origin'])[measure].mean()
        .unstack()
        .plot(figsize=(6, 4), title=f'Mean {measure} by model year'))
    plt.show()
```

After the oil crisis of 1973, mean mileage rose in all three regions.

American cars doubled theirs, from about 15 to 30, while losing over 1,000 pounds.

## Which measures go together?

Pick two measures to plot against each other.

```python
@interact(x=measure_cols, y=measure_cols)
def plot_scatter(x='weight', y='mpg'):
    MPG.plot.scatter(x=x, y=y, c='origin', figsize=(6, 4))
    plt.show()
```
